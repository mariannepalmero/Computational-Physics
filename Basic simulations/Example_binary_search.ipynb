### Example Binary Search

Solve the equation $f(x) = e^{-x} - x = 0$ using binary search in interval $[0,1]$ with an accuracy of $10^{-6}$. 

```python
from math import exp

# Constants
accuracy = 10 ** -6
x1, x2 = 0, 1 # Initial interval
iterations = 0

while True:
    x = (x1 + x2) / 2  # Midpoint

    # Functions
    f_x1 = exp(-x1) - x1
    f_x = exp(-x) - x
    f_x2 = exp(-x2) - x2
    
    # If f_x1 and f_x have the same sign
    if (f_x1 < 0 and f_x < 0) or (f_x1 > 0 and f_x > 0): 
        x1 = x

    # If f_x and f_x2 have the same sign
    elif (f_x > 0 and f_x2 > 0) or (f_x < 0 and f_x2 < 0):
        x2 = x
        
    if abs(x1 - x2) < accuracy:
        break
    iterations += 1

x_solution = (x1 + x2) / 2

print(f"Solution: {x}, Last Interval ({x1}, {x2}), Iterations: {iterations}")
```
