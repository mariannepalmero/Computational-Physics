### Example:

Find a root of the equation $f(x) = x^{3} -5x +1 = 0$ using secant method, starting from $x_{0} = 0$ and $x_{1} = 1$.

```python
#Constants
accuracy = 10 ** -4

def f(x):
    return x**3 - 5*x + 1

def secant_method(f,accuracy):
    """
    Compute the f(x) using the secant method.
    param f (float): The answer from f function.
    param accuracy (float): The accuracy tolerance.
    return x (float): The computed f(x).
    """
    # Initial guesses
    x0 = 0 
    x1 = 1
    
    while True:
        f_x0 = f(x0)
        f_x1 = f(x1)
        x_new = x1 - f_x1 * (x1 - x0) / (f_x1 - f_x0)
        if abs(x_new - x1) < accuracy:
            break
        x0, x1 = x1, x_new
    
    return x1

x_solution = secant_method(f, accuracy)

print(f"The solution is x = {x_solution:.4f}.")
```
