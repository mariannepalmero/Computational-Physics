### **Exercise 6.13: Wien's displacement constant**

Planck's radiation law tells us that the intensity of radiation per unit area and per unit wavelength, $\lambda$ from a black body at temperature $T$ is
$I (\lambda) = \frac{2 \pi h c^{2} \lambda^{-5}}{e^{\frac{hc}{\lambda k_{B} T}} - 1}$, where $h$ is Planck's constant, $c$ is the speed of light, and $k_{B}$ is Boltzmann's constant.

a) Show by differentiating that the wavelength $\lambda$ at which the emitted radiation is strongest is the solution of the equation $5e^{- \frac{hc}{\lambda k_{B} T}} + \frac{hc}{\lambda k_{B} T} - 5 = 0$. Make the substitution $x = \frac{hc}{\lambda k_{B} T}$ and hence show that the wavelength of maximum radiation obeys the *Wien displacement law* $\lambda = \frac{b}{T}$, where the so-called *Wien displacement constant* is $b = \frac{hc}{k_{B} x}$, and $x$ is the solution to the nonlinear equation $5e^{-x} + x - 5 = 0$.

b) Write a program to solve this equation to an accuracy of $\epsilon = 10^{-6}$ using the binary search method, and hence find a value for the displacement constant.

c) The displacement law is the basis for the method of *optical pyrometry*, a method for measuring the temperatures of objects by observing the color of the thermal radiation they emit. The method is commonly used to estimate the surface temperatures of astronomical bodies, such as the Sun. The wavelength peak in the Sun's emitted radiation falls at $\lambda = 502$ nm. From the equations above and your value of the displacement constant, estimate the surface temperature of the Sun.

a) We are given the Planck's radiation law $I (\lambda) = \frac{2 \pi h c^{2} \lambda^{-5}}{e^{\frac{hc}{\lambda k_{B} T}} - 1}$. We need to find the wavelength $\lambda$ at which the intensity is maximum. We differentiate $I(\lambda)$ with respect to $\lambda$ and set the derivative to $0$: $\frac{dI}{d \lambda} = 0$.

We differentiate using the quotient rule:

$\frac{dI}{d \lambda} = \frac{d}{d \lambda}(\frac{2 \pi h c^{2} \lambda^{-5}}{e^{x} - 1})$ 

$\frac{dI}{d \lambda} = \frac{(2 \pi h c^{2} \lambda^{-5})'(e^{\frac{hc}{\lambda k_{B} T}} - 1)-(2 \pi h c^{2} \lambda^{-5})(e^{\frac{hc}{\lambda k_{B} T}} - 1)'}{(e^{\frac{hc}{\lambda k_{B} T}} - 1)^2}$

$\frac{dI}{d \lambda} = \frac{(-5 \frac{2 \pi h c^{2}}{lambda ^{6}})(e^{\frac{h c}{\lambda k_{B} T}} - 1)-(\frac{2 \pi h c^{2}}{\lambda^{5}})[e^{\frac{hc}{\lambda k_{B} T}}(- \frac{hc}{\lambda k_{B} T})]}{(e^{\frac{hc}{\lambda k_{B} T}}-1)^{2}}$

We let $x = \frac{hc}{\lambda k_{B} T}$. So, $e^{x} = e^{\frac{hc}{\lambda k_{B} T}}$.

$\frac{dI}{d \lambda} = \frac{(-5 \frac{2 \pi h c^{2}}{\lambda ^{6}})(e^{x} - 1)-(\frac{2 \pi h c^{2}}{\lambda^{5}})[e^{x}(- \frac{x}{\lambda})]}{(e^{x}-1)^{2}}$

$\frac{dI}{d \lambda} = \frac{\frac{2 \pi h c^{2}}{\lambda^{6}} [-5 (e^{x} -1) + x e^{x}]}{(e^{x}-1)^2}$

We set the derivative to $0$: $\frac{dI}{d \lambda} = 0$ to find the wavelength $\lambda$ at which the intensity is maximum.

$-5 (e^{x} -1) + x e^{x} = 0$

$-5e^{x} + 5 + xe^{x} = 0$

$\fbox{$5 e^{x} + x - 5 = 0$}$

b) We use $[4, 6]$ as initial interval; $f(4)<0$ and $f(6)>0$.

```python
from math import exp

# Constants
accuracy = 10 ** -6
x1, x2 = 4, 6 # Initial interval
iterations = 0

# Physical Constants
h = 6.62607015e-34  # Planck's constant (J·s)
c = 2.99792458e8    # Speed of light (m/s)
k_B = 1.380649e-23  # Boltzmann's constant (J/K)

while True:
    x = (x1 + x2) / 2  # Midpoint

    # Functions
    f_x1 = 5 * exp(-x1) + x1 - 5
    f_x = 5 * exp(-x) + x - 5
    f_x2 = 5 * exp(-x2) + x2 - 5
    
    # If f_x1 and f_x have the same sign
    if (f_x1 < 0 and f_x < 0) or (f_x1 > 0 and f_x > 0): 
        x1 = x

    # If f_x and f_x2 have the same sign
    elif (f_x > 0 and f_x2 > 0) or (f_x < 0 and f_x2 < 0):
        x2 = x
        
    if abs(x1 - x2) < accuracy:
        break
        
    iterations += 1

x_solution = (x1 + x2) / 2

# Compute Wien's displacement constant
b = h * c / (k_B * x_solution)

print(f"Wien's displacement constant: {b} m·K, Solution: {x}, Last Interval ({x1}, {x2}), Iterations: {iterations}")
```

c) We are given the wavelength at which the radiation from the Sun is strongest $\lambda_{max} = 502$ nm $= 502 \times 10^{-9}$ m. We need to determine the estimate of the surface temperature of the Sun. 

We use Wien's displacement law $\lambda_{max} = \frac{b}{T}$. Rearranging, we have $T = \frac{b}{\lambda_{max}}$.

```python
# Wavelength at which the radiation from the Sun is strongest
wavelength_max = 502 * 10 ** -9

# Formula for the temperature
T = b / wavelength_max

print(f"The estimate surface temperature of the Sun is: {T} K.")
```
