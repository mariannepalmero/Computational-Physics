### Example Relaxation Method: Ferromagnetism

In the mean-field theory of ferromagnetism, the strength $M$ of magnetization of a ferromagnetic material like iron depends on temperature $T$ according to the formula $M = \mu \tanh \frac{JM}{k_{B}T}$, where $\mu$ is a magnetic moment, $J$ is a coupling constant, and $k_{B}$ is the Boltzmann's constant. Simplifying, we let $m = \frac{M}{\mu}$ and $C = \frac{\mu J}{k_{B}}$ so that $m = \tanh \frac{Cm}{T}$. 

This equation clearly has a solution at $m = 0$. However, we should also check if there are solutions $m \neq 0$. We assume that $C=1$ and look for solutions as a function of $T$, accurate within $\pm 10^{-6}$ of the true answer.
