### Exercise 6.16: The Lagrange point
There is a magical point between the Earth and the Moon, called the $L_{1}$ Lagrange point, at which a satellite will orbit the Earth in perfect synchrony with the Moon, staying always in between the two. This works because the inward pull of the Earth and the outward pull of the Moon combine to create exactly the needed centripetal force that keeps the satellite in its orbit.

a) Assuming circular orbits, and assuming that the Earth is much more massive than either the Moon or the satellite, show that the distance r from the center of the Earth to the $L_{1}$ point satisfies 

$\frac{GM}{r^{2}} - \frac{Gm}{(R - r)^{2}} = \omega ^2 r$

where $R$ is the distance from the Earth to the Moon, $M$ and $m$ are the Earth and Moon masses, $G$ is Newton's gravitational constant, and $\omega$ is the angular velocity of both the Moon and the satellite.

b) The equation above is a degree-five polynomial equation in $r$ (also called a quintic equation). Such equations cannot be solved exactly in closed form, but it's straightforward to solve them numerically. Write a program that uses either Newton's method or the secant method to solve for the distance $r$ from the Earth to the $L_1$ point. Compute a solution accurate to at least four significant figures.

a) We are given the location of satellite $L_{1}$.

We ned to show that the distance r from the e center of the Earth to the $L_{1}$ point satisfies 

$\frac{GM}{r^{2}} - \frac{Gm}{(R - r)^{2}} = \omega ^2 r$

where $R$ is the distance from Earth to moon, $M$ is the mass of Earth, $m$ is the mass of Moon, $G$ is the Newton's gravitational constant, and $\omega$ is the angular velocity of Moon and satellite.

We analyze the forces acting on a satellite located at $L_{1}$ to find the distance $r$ from Earth to $L_{1}$.

The forces acting on satellite at $L_{L}$ are:
- the gravitational attraction of the Earth: $F_{E} = \frac{GMm_{s}}{r^{2}}$
- the gravitational attraction of the Moon: $F_{M} = \frac{Gmm_{s}}{(R-r)^{2}}$

where $m_{s}$ is the mass of the satellite.

The satellite is in a rotating reference frame where it maintains a fixed position relative to the Earth and Moon. It experiences a centripetal acceleration $\omega^{2} r$. So the net radial force acting on the satellite is:

$F_{E} - F_{M} = m_{s} \omega^{2} r$

$\frac{GMm_{s}}{r^{2}} - \frac{Gmm_{s}}{(R-r)^{2}} = m_{s} \omega^{2} r$

Simplifying, we have
$\fbox{$\frac{GM}{r^{2}} - \frac{Gm}{(R-r)^{2}} = \omega^{2} r$}$

b) **Newton's Method**

We have the equation $\frac{GM}{r^{2}} - \frac{Gm}{(R - r)^{2}} = \omega ^2 r$. Rewriting this,  $r$ is a root of the equation $\frac{GM}{r^{2}} - \frac{Gm}{(R - r)^{2}} - \omega ^2 r = 0$. 

Solving the equation $x'=x - \Delta x = x - \frac{f(x)}{f'(x)}$ for this problem, we have $r' = r - \frac{\frac{GM}{r^{2}} - \frac{Gm}{(R - r)^{2}} - \omega ^2 r}{\frac{-2GM}{r^{3}} + \frac{2Gm}{(R-r)^3} - \omega^2}$.

We start with initial guess $r=\frac{R}{2}$. The iterative formula we will use is $r' = r - \frac{\frac{GM}{r^{2}} - \frac{Gm}{(R - r)^{2}} - \omega ^2 r}{\frac{-2GM}{r^{3}} + \frac{2Gm}{(R-r)^3} - \omega^2}$.

```python
#Constants
G = 6.674 * 10 ** -11  # gravitational constant
M = 5.974 * 10 ** 24  # mass of earth in kg
m = 7.348 * 10 ** 22  # mass of moon in kg
R = 3.844 * 10 ** 8  # Radius of earth in m
omega = 2.662 * 10 ** -6  # angular velocity of moon
accuracy = 10 ** -4

def f(r, G, M, m, R, omega):
    return (G * M / r**2) - (G * m / (R - r)**2) - (omega**2 * r)

def f_prime(r, G, M, m, R, omega):
    return (-2 * G * M / r**3) + (2 * G * m / (R - r)**3) - omega**2

def newtons_method(f,f_prime,accuracy, G, M, m, R, omega):
    """
    Compute the f(r) using Newton's method.
    param f (float): The answer from f function.
    param f_prime (float): The answer from f_prime function.
    param accuracy (float): The accuracy tolerance.
    param G (float): The gravitational constant.
    param M (float): The mass of earth.
    param m (float): The mass of moon.
    param R (float): The radius of earth.
    param omega (float): The angular velocity of moon.
    return r (float): The computed f(r).
    """
    r= R / 2 # Initial guess
    while True:
        r_new = r - f(r, G, M, m, R, omega) / f_prime(r, G, M, m, R, omega)
        if abs(r_new - r) < accuracy:
            break
        r = r_new
    return r

r_solution = newtons_method(f, f_prime, accuracy, G, M, m, R, omega)

print(f"The L1 point is located at r = {r_solution:.4f} meters from Earth's center.")
```

b) **Secant Method**

We start with initial guesses $r_0=\frac{R}{3}$ and $r_1=\frac{R}{2}$.  For $f(r) = \frac{GM}{r^{2}} - \frac{Gm}{(R - r)^{2}} - \omega ^2 r$, the iterative formula we will use is $r_2 = r_1 - f(r_1) * \frac{r_1 - r_0}{f(r_1) - f(r_0)}$.

```python
#Constants
G = 6.674 * 10 ** -11  # gravitational constant
M = 5.974 * 10 ** 24  # mass of earth in kg
m = 7.348 * 10 ** 22  # mass of moon in kg
R = 3.844 * 10 ** 8  # Radius of earth in m
omega = 2.662 * 10 ** -6  # angular velocity of moon
accuracy = 10 ** -4

def f(r, G, M, m, R, omega):
    return (G * M / r**2) - (G * m / (R - r)**2) - (omega**2 * r)

def secant_method(f,accuracy, G, M, m, R, omega):
    """
    Compute the f(r) using secant method.
    param f (float): The answer from f function.
    param accuracy (float): The accuracy tolerance.
    param G (float): The gravitational constant.
    param M (float): The mass of earth.
    param m (float): The mass of moon.
    param R (float): The radius of earth.
    param omega (float): The angular velocity of moon.
    return r (float): The computed f(r).
    """
    # Initial guesses
    r0 = R / 3 
    r1 = R / 2
    
    while True:
        f_r0 = f(r0, G, M, m, R, omega)
        f_r1 = f(r1, G, M, m, R, omega)
        r_new = r1 - f_r1 * (r1 - r0) / (f_r1 - f_r0)
        if abs(r_new - r1) < accuracy:
            break
        r0, r1 = r1, r_new
    
    return r1

r_solution = secant_method(f, accuracy, G, M, m, R, omega)

print(f"The L1 point is located at r = {r_solution:.4f} meters from Earth's center.")
```
