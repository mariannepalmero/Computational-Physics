### Example: Inverse Hyperbolic Tangent

Let us use Newton's method to calculate the inverse (or arc) hyperbolic tangent of a number $\mu$. 

By definition, $\arctan h \mu$ is the number $x$ such that $\mu = \tanh x$. To put that another way, $x$ is a root of the equation $\tanh x - \mu = 0$. 

Recalling that the derivative of $\tanh x$ is $\frac{1}{\cosh^{2}x}$, equation $x'=x - \Delta x = x - \frac{f(x)}{f'(x)}$ for this problem becomes $x' = x - (\tanh x - \mu) \cosh^{2} x$.

We start with initial guess $x=0$. The iterative formula we will use is $x' = x - (\tanh x - \mu) \cosh^{2} x$

```python
from math import tanh, cosh
from numpy import linspace 
from pylab import plot,show,xlabel,ylabel

#Constants
accuracy = 10 ** -12

def f(x, mu):
    return (tanh(x)-mu)

def f_prime(x, mu):
    return (1/cosh(x)**2)

def newtons_method(mu, f, f_prime, accuracy): 
    """
    Compute the f(x, mu) using Newton's method.
    param mu (float): The input value, must be in the range (-1, 1).
    param f (float): The answer for f function.
    param f_prime (float): The answer for f_prime function.
    param accuracy (float): The accuracy tolerance.
    return x (float): The computed arctanh(mu).
    """
    x = 0.0 # Initial guess

    while True:
        x_new = x - f(x, mu) / f_prime(x, mu)
        if abs(x_new - x) < accuracy:
            break
        x = x_new
    return x
    

# Make array of x and mu
mupoints = linspace(-0.99,0.99,100) 
xpoints = []


for mu in mupoints: 
    xpoints.append(newtons_method(mu, f, f_prime, accuracy))

# Make the graph
plot(mupoints,xpoints) 
xlabel("$\mu$") 
ylabel("x")
show()
```
